```python
import numpy as np
import pandas as pd
df = pd.read_csv('titanic.csv')
df.head()
```

```python
df['number'].unique()
```

```python
fig=df['number'].value_counts().plot(kind='bar',color='red')
```

```python
df['numeric_numbers']=pd.to_numeric(df['number'],errors='coerce',downcast='integer')
df['numeric_numbers'].unique()
```

```python
df.head()
```

```python
df['cat_numbers']=np.where(df['numeric_numbers'].isnull(),df['number'],np.nan)
df.head()
```

```python
df['Cabin'].unique()
```

```python
df['cabin_num'] = df['Cabin'].str.extract('(\d+)') # captures numerical part
df['cabin_cat'] = df['Cabin'].str[0] # captures the first letter

df.head()
```

```python
df['cabin_cat'].value_counts().plot(kind='bar')
```

```python
# extract the last bit of ticket as number
df['ticket_num'] = df['Ticket'].apply(lambda s: s.split()[-1])
df['ticket_num'] = pd.to_numeric(df['ticket_num'],
                                   errors='coerce',
                                   downcast='integer')

# extract the first part of ticket as category
df['ticket_cat'] = df['Ticket'].apply(lambda s: s.split()[0])
df['ticket_cat'] = np.where(df['ticket_cat'].str.isdigit(), np.nan,
                              df['ticket_cat'])

df.head(20)
```
