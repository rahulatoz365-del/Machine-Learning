```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

**Load the datasets(Multiple Datasets are used)**

```python
train=pd.read_csv("train.csv")
train.head()
```

```python
tips=sns.load_dataset("tips")
tips.head()
```

```python
flights=sns.load_dataset("flights")
flights.head()
```

```python
iris=sns.load_dataset("iris")
iris.head()
```

**1. Scatterplot(Numerical-Numerical)**

**Bi Variate analysis**

```python
sns.scatterplot(data=tips,x="total_bill",y="tip")
```

**Multi Variate analysis**

```python
sns.scatterplot(data=tips,x="total_bill",y="tip",hue="sex",style="smoker",size="size")
```

**2. Bar Plot (Numerical-Categorical)**

**Bi Variate analysis**

```python
sns.barplot(data=train,x="Pclass",y="Survived")
```

**Multi Vairate analysis**

```python
sns.barplot(data=train,x="Pclass",y="Survived",hue="Sex")
```

**3. Box Plot (Numerical-Categorical)**

**Bi Variate Analysis**

```python
sns.boxplot(data=train,x="Pclass",y="Age")
```

**Multi Variate Analysis**

```python
sns.boxplot(data=train,x="Sex",y="Age",hue="Survived")
```

**4. Dist Plot(Numerical-Categorical)**

**Bi Variate analysis**

```python
sns.kdeplot(data=train,x="Age",hue="Survived")
```

**5. HeatMap(Categorical-Categorical)**

**Bi Variate Analysis**

```python
sns.heatmap(pd.crosstab(train["Pclass"],train["Survived"]),annot=True,fmt="d")
```

**6. Clustermap**

```python
sns.clustermap(pd.crosstab(train["Pclass"],train["Survived"]),annot=True,fmt="d")
```

**7. Pair Plot**

```python
sns.pairplot(iris,hue="species")
```

**8. Line Plot**

```python
new = flights.groupby('year').sum(numeric_only=True).reset_index()
```

```python
sns.lineplot(data=new, x='year', y='passengers')
```
