```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
```

```python
X,y = make_regression(n_samples=100, n_features=1, n_informative=1, n_targets=1,noise=20,random_state=13)
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression, Ridge
model = LinearRegression()
model.fit(X,y)
model.coef_, model.intercept_
```

```python
ridge_model = Ridge(alpha=10)
ridge_model.fit(X,y)
ridge_model.coef_, ridge_model.intercept_
```

```python
ridge_model_1 = Ridge(alpha=100)
ridge_model_1.fit(X,y)
ridge_model_1.coef_, ridge_model_1.intercept_
```

```python
plt.scatter(X,y)
plt.plot(X, model.predict(X), color='red', label='Linear Regression')
plt.plot(X, ridge_model.predict(X), color='green', label='Ridge Regression (alpha=10)')
plt.plot(X, ridge_model_1.predict(X), color='blue', label='Ridge Regression (alpha=100)')
plt.legend()
plt.show()
```

```python
def linear_regression(X, y , alpha=1.0):
    x_mean = X.mean()
    y_mean = y.mean()
    numerator=0
    denominator=0
    for i in range(len(X)):
        numerator += (X[i] - x_mean) * (y[i] - y_mean)
        denominator += (X[i] - x_mean) ** 2
    m = numerator / denominator*alpha #Slope
    c = y_mean - m * x_mean #Intercept
    return m, c
```

```python
class RidgeRegressionScratch:
    def __init__(self, alpha=1.0):
        self.alpha = alpha
        self.coef_ = None
        self.intercept_ = None

    def fit(self, X, y):
        num=0
        den=0
        for i in range(len(X)):
            num += (X[i] - X.mean()) * (y[i] - y.mean())
            den += (X[i] - X.mean()) ** 2
        self.coef_ = num / (den + self.alpha) #Slope
        self.intercept_ = y.mean() - self.coef_ * X.mean() #Intercept
        print(f"Coefficient: {self.coef_}, Intercept: {self.intercept_}")
```

```python
RidgeRegressionScratch(alpha=10).fit(X,y)
```
