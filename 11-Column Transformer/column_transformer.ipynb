```python
import numpy as np
import pandas as pd
```

```python
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
df=pd.read_csv('covid_toy.csv')
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df.drop('has_covid',axis=1),df['has_covid'],test_size=0.2,random_state=42)
X_train
```

**Tedious Way of Data Preprocessing(Without Column Transformer)**

```python
imputer=SimpleImputer(strategy='median')
X_train_fever=imputer.fit_transform(X_train[['fever']])
X_test_fever=imputer.transform(X_test[['fever']])
X_train_fever
```

```python
df['cough'].value_counts()
```

```python
oe=OrdinalEncoder(categories=[['Mild'  ,'Strong']])
X_train_cough=oe.fit_transform(X_train[['cough']])
X_test_cough=oe.transform(X_test[['cough']])
X_train_cough
```

```python
ohe=OneHotEncoder(sparse_output=False,drop='first')
X_train_gender_city=ohe.fit_transform(X_train[['gender','city']])
X_test_gender_city=ohe.transform(X_test[['gender','city']])
X_train
```

```python
X_train_age=X_train[['age']].values
X_test_age=X_test[['age']].values
X_train_age
```

```python
X_train_transformed = np.concatenate((X_train_age,X_train_fever,X_train_gender_city,X_train_cough),axis=1)
X_test_transformed = np.concatenate((X_test_age,X_test_fever,X_test_gender_city,X_test_cough),axis=1)
X_train_transformed.shape
```

**Easy Process with the help of Column Transformer**

```python
from sklearn.compose import ColumnTransformer
```

```python
transformer=ColumnTransformer(transformers=[
    ('fever_imputer',SimpleImputer(strategy='median'),['fever']),
    ('cough_encoder',OrdinalEncoder(categories=[['Mild'  ,'Strong']]),['cough']),
    ('gender_city_encoder',OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
],remainder='passthrough')
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test).shape
```
