```python
import numpy as np
from sklearn.datasets import load_diabetes
```

```python
X,y = load_diabetes(return_X_y=True)
X.shape,y.shape
```

**Making Own Multiple Linear Regression Class from Scratch (This would also supported single/multiple Linear Regression both)**

```python
class Multi_LR:
    def __init__(self):
        self.m = None #y=mx+c --> m is the slope
        self.c = None #y=mx+c --> c is the intercept
    
    def fit(self,X,y):
        X = np.insert(X,0,1,axis=1) #inserting a column of 1's for the intercept
        beta = np.linalg.inv(np.dot(X.T, X)).dot(X.T).dot(y) #calculating the coefficients using the normal equation
        self.c = beta[0] #intercept
        self.m = beta[1:] #slopes(coefficients)

    def predict(self,X_test):
        y_pred = self.c + np.dot(X_test,self.m) #y=mx+c
        return y_pred
```

```python
lr = Multi_LR()
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=3)
lr.fit(X_train,y_train)
X_train.shape,y_train.shape
```

```python
np.insert(X_train,0,1,axis=1).shape
```

```python
y_pred = lr.predict(X_test)
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
print("MSE=",mean_squared_error(y_test,y_pred))
print("R2 Score=",r2_score(y_test,y_pred))
print("MAE=",mean_absolute_error(y_test,y_pred))
print("RMSE=",np.sqrt(mean_squared_error(y_test,y_pred)))
```
