```python
import numpy as np
import pandas as pd
```

```python
df= pd.read_csv('cars.csv')
df.sample(10)
```

```python
df['owner'].value_counts()
```

**1. One Hot Encoding using Pandas**

```python
#Dummy Variable Trap
pd.get_dummies(df,columns=['owner','fuel'])
```

**2. K-1 One Hot Encoding OR N-1 One Hot Encoding**

```python
#Get rid of one dummy variable to avoid Dummy Variable Trap
pd.get_dummies(df,columns=['owner','fuel'],drop_first=True)
```

**4. One Hot Encoding with Top Categories**

```python
counts=df['brand'].value_counts()
```

```python
df['brand'].nunique()
threshold=100
```

```python
repl=counts[counts<=threshold].index
```

```python
pd.get_dummies(df['brand'].replace(repl,'uncommon')).sample(10)
```
